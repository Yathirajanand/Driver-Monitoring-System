```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8m.pt")
```

```python
model.train(
    data="driver_dataset/data.yaml",
    epochs=120,
    imgsz=640,
    batch=16,
    patience=30,      # prevents overfitting
    name="driver_monitoring_high_accuracy"
)
```

```python
model = YOLO("runs/detect/driver_monitoring_6class/weights/best.pt")

model.predict(
    source="/content/Driver2.jpg",
    conf=0.4,
    save=True
)
```

```python
results = model("Driver2.jpg", conf=0.4)

for r in results:
    for box in r.boxes:
        label = model.names[int(box.cls)]
        print("Detected:", label)
        if label == "hold_phone" or label == "phone":
            print("📵 ALERT: Phone detected!")
        if label == "cigarette":
            print("🚬 ALERT: Smoking!")
        if label == "EyeClose":
            print("😴 ALERT: Eyes closed!")
```
